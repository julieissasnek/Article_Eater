## 3. Start the API and worker

This cell starts the FastAPI app and the Article Eater worker in the background.
On hosted environments you may see logs in the notebook output.


````python
import subprocess, sys, os, time

# Start API
api_cmd = [sys.executable, "-m", "uvicorn", "app.main:app", "--host", "0.0.0.0", "--port", "8000"]
api_proc = subprocess.Popen(api_cmd)
print("Started API on http://0.0.0.0:8000 (PID)", api_proc.pid)

# Start worker
worker_cmd = [sys.executable, "app/worker.py"]
worker_proc = subprocess.Popen(worker_cmd)
print("Started worker (PID)", worker_proc.pid)

# Give them a few seconds to boot
time.sleep(5)
````

## 4. Start the Streamlit Control Room

This cell launches the Streamlit control room. In some hosted environments, you may
get a public URL automatically or you may need to follow the link in the cell output.


````python
import subprocess, sys

streamlit_cmd = [
    sys.executable,
    "-m",
    "streamlit",
    "run",
    "scripts/ae_streamlit_control_room.py",
    "--server.port",
    "8501",
    "--server.address",
    "0.0.0.0",
]
st_proc = subprocess.Popen(streamlit_cmd)
print("Started Streamlit control room on port 8501 (PID)", st_proc.pid)
print("\nIf you're in Colab or Codespaces, use the provided URL or open a forwarded port.")
````
